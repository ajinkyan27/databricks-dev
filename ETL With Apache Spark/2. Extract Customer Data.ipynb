 2. **Query multiple JSON files**

Select File Metadata

```python
%sql
SELECT *,input_file_name() as File_Path,_metadata.file_path as FilePath FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers/`
```

## Create Views

```python
%sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers
AS
SELECT *,input_file_name() as File_Path,_metadata.file_path as FilePath FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers/`
```

```python
%sql
SELECT * FROM gizmobox.bronze.v_customers
```
